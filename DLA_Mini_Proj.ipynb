Import Libraries

```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, SimpleRNN, LSTM, Conv1D, GlobalMaxPooling1D
```

Load Dataset

```python
vocab_size = 10000

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=vocab_size)
```

Preprocess Data

```python
max_len = 200

x_train = pad_sequences(x_train, maxlen=max_len)
x_test = pad_sequences(x_test, maxlen=max_len)
```

CNN Model

```python
cnn_model = Sequential()

cnn_model.add(Embedding(vocab_size, 128, input_length=max_len))
cnn_model.add(Conv1D(64, 5, activation='relu'))
cnn_model.add(GlobalMaxPooling1D())
cnn_model.add(Dense(10, activation='relu'))
cnn_model.add(Dense(1, activation='sigmoid'))

cnn_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

cnn_model.fit(x_train, y_train, epochs=3, batch_size=64)
```

RNN Model

```python
rnn_model = Sequential()

rnn_model.add(Embedding(vocab_size, 128, input_length=max_len))
rnn_model.add(SimpleRNN(64))
rnn_model.add(Dense(1, activation='sigmoid'))

rnn_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

rnn_model.fit(x_train, y_train, epochs=3, batch_size=64)
```

LSTM Model

```python
lstm_model = Sequential()

lstm_model.add(Embedding(vocab_size, 128, input_length=max_len))
lstm_model.add(LSTM(64))
lstm_model.add(Dense(1, activation='sigmoid'))

lstm_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

lstm_model.fit(x_train, y_train, epochs=3, batch_size=64)
```

Model Comparison

```python
cnn_acc = cnn_model.evaluate(x_test, y_test)[1]
rnn_acc = rnn_model.evaluate(x_test, y_test)[1]
lstm_acc = lstm_model.evaluate(x_test, y_test)[1]

print("CNN Accuracy:", cnn_acc)
print("RNN Accuracy:", rnn_acc)
print("LSTM Accuracy:", lstm_acc)
```
